## 这个 notebook 在做什么

- 数据集：`red_train_yolo.zip` 里是 61 张训练图 + 13 张验证图，标注只有一类 `red_block`。
- 红块在原图里只有 **9–25 像素**，所以 `imgsz=1280`；降到 640 会把块缩成几个像素，学不到。
- 训练产出两个文件：`best.pt`（PyTorch 格式，以后继续训练用）和 `best.onnx`（给车上跑推理用）。

```python
!unzip -q -o red_train_yolo.zip -d /content/data
!ls /content/data
import glob, yaml
path = '/content/data/data.yaml'
cfg = yaml.safe_load(open(path))
cfg['path'] = '/content/data'          # data.yaml 里记的是 Windows 绝对路径
yaml.safe_dump(cfg, open(path, 'w'), allow_unicode=True)
print(cfg)
print('train images', len(glob.glob('/content/data/images/train/*.jpg')))
print('val images', len(glob.glob('/content/data/images/val/*.jpg')))
```

```python
from ultralytics import YOLO

model = YOLO('yolov8n.pt')
result = model.train(
    data='/content/data/data.yaml',
    imgsz=1280,          # 块只有 10 多像素，别用 640
    epochs=120,
    batch=8,
    patience=30,
    workers=2,
    project='/content/runs',
    name='red_block',
    exist_ok=True,
    scale=0.6,
    translate=0.15,
    degrees=10.0,
    hsv_h=0.015,
    hsv_s=0.7,
    hsv_v=0.4,
    mosaic=1.0,
    close_mosaic=20,
)
```

```python
# 验证：mAP 之外重点看 recall（漏检）
metrics = model.val(data='/content/data/data.yaml', imgsz=1280, split='val')
print('mAP50   ', metrics.box.map50)
print('mAP50-95', metrics.box.map)
print('precision', metrics.box.mp)
print('recall   ', metrics.box.mr)
```

```python
# 挑几张验证图看实际效果（比数字更可靠）
import glob
val_images = sorted(glob.glob('/content/data/images/val/*.jpg'))[:6]
for path in val_images:
    model.predict(path, imgsz=1280, conf=0.25, save=True, project='/content/runs/predict', name='val')
print('结果在 /content/runs/predict/val')
```

```python
# 导出 ONNX（固定 1280x960，和取流帧一致），并下载到本机
best = '/content/runs/red_block/weights/best.pt'
m = YOLO(best)
onnx_path = m.export(format='onnx', imgsz=(960, 1280), opset=12, simplify=True)
print(onnx_path)
files.download(onnx_path)
files.download(best)
```

## 判读标准

| 指标 | 期望 | 说明 |
|---|---|---|
| recall | ≥ 0.90 | 漏检比误检更伤；比赛少看到一个块就会推空 |
| mAP50 | ≥ 0.90 | 有 1326 个框，应该容易到 |
| 实际图 | 聚堆能分开 | 只看数字会漏掉“一堆被框成一条”的问题 |
